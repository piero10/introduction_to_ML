## Домашнее задание
В данном задании вам предстоит реализовать сервис на FastAPI который будет включать в себя построенную модель машинного обучения.

### Замечание:
1. Ваше решение может включать в себя только следующие библиотеки
        - fastapi==0.116.1
        - uvicorn[standard]==0.23.2
        - pydantic==2.12.3
        - scikit-learn==1.3.3
        - pandas==2.1.1
Если вы очень хотите использовать какие-то дополнительные библиотеки, то реализуйте ваше решение в докер
т.е. дорлжен быть docker- файл + readme как его запустить, к каким его использовать, к каким эндпоинтам обращаться


### Задания:
1. Постройте модель по определению "ядовитости" грибов из 3-его урока.
2. Сериализуйте модель в pkl файл
3. Напишите сервис с использованием FastAPI или Flask который должен реализовывать следующие get эндпоиты:
    - predict (get) - передаем данные одного гриба, надо определить ядовитый он или нет (с использованием модели)
    - predict_proba - передаем данные одного гриба, необходимо определить вероятность того, что гриб ядовитый
    - predict_batch - передаем данные нескольких грибов, надо определить ядовитый ли каждый из них
    - predict_proba_batch - .... аналогично
    - status - возвращает даты когда модель была обучене (загружена из pkl фафла)

4. Реализуйте следующие post эндпоинты:
    - fit - передаем датафрейм с данными грибов, модель строиться заново, обучается на принятых данных, сериализуется в pkl файл

5. Напишите хотя бы один тест для сервера  

6. Напишите клиент, который будет отправлять запросы всех перечисленных выше типов на сервер

8. При реализации проекта старайтесь соблюдайте все стандарты качества кода:
    - pep8
    - precommiter
    - аннатацияr
    - pydantic
    - тесты для сервис всего 5 тестов для get методов
    
9. Результат вашей работы должен включать в себя:
    - код, выложите ваш код на любое открытое git хранилище
    - readme.md файл, который содержит в себе следующее инфо:
       - как запустить сервер (команда, ссылка на bat / sh файл)
       - примеры url с вызовом predict, predict_proba, ... все методы которые вам удалось реализовать
