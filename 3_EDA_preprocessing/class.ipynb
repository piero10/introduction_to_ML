### Фича инженеринг

```python
def engineer_features(df):
    """
    Create new features based on domain knowledge and data exploration
    """
    df_new = df.copy()
    
    # Total square footage
    df_new['TotalSF'] = df_new['TotalBsmtSF'] + df_new['1stFlrSF'] + df_new['2ndFlrSF']
    
    # Total bathrooms
    df_new['TotalBathrooms'] = (df_new['FullBath'] + 
                               0.5 * df_new['HalfBath'] +
                               df_new['BsmtFullBath'] + 
                               0.5 * df_new['BsmtHalfBath'])
    
    # Total porch area
    df_new['TotalPorchSF'] = (df_new['OpenPorchSF'] + 
                             df_new['3SsnPorch'] +
                             df_new['EnclosedPorch'] + 
                             df_new['ScreenPorch'] +
                             df_new['WoodDeckSF'])
    
    # House age
    df_new['HouseAge'] = df_new['YrSold'] - df_new['YearBuilt']
    df_new['RemodAge'] = df_new['YrSold'] - df_new['YearRemodAdd']
    
    # Quality metrics
    df_new['OverallGrade'] = df_new['OverallQual'] * df_new['OverallCond']
    df_new['QualityArea'] = df_new['OverallQual'] * df_new['GrLivArea']
    
    # Binary features
    df_new['HasBasement'] = (df_new['TotalBsmtSF'] > 0).astype(int)
    df_new['HasGarage'] = (df_new['GarageArea'] > 0).astype(int)
    df_new['HasFireplace'] = (df_new['Fireplaces'] > 0).astype(int)
    df_new['HasPool'] = (df_new['PoolArea'] > 0).astype(int)
    df_new['Has2ndFloor'] = (df_new['2ndFlrSF'] > 0).astype(int)
    
    # Ratios and interactions
    df_new['LivAreaRatio'] = df_new['GrLivArea'] / df_new['TotalSF']
    df_new['BsmtRatio'] = df_new['TotalBsmtSF'] / df_new['TotalSF']
    
    return df_new

# Apply feature engineering
train_fe = engineer_features(df)
test_fe = engineer_features(df_test)

print(f"Original features: {df.shape[1]}")
print(f"After feature engineering: {train_fe.shape[1]}")
```

### Построение модели

собираем все данные вместе

```python
def preprocess_data(train_df, test_df):
    """
    Preprocess data for modeling
    """
    # Combine datasets for consistent preprocessing
    all_data = pd.concat([train_df, test_df], sort=False)
    
    # Handle missing values
    # Numerical features - fill with median
    numeric_features = all_data.select_dtypes(include=[np.number]).columns
    for col in numeric_features:
        if col != 'SalePrice':
            all_data[col] = all_data[col].fillna(all_data[col].median())
    
    # Categorical features - fill with mode or 'None'
    categorical_features = all_data.select_dtypes(include=['object']).columns
    for col in categorical_features:
        all_data[col] = all_data[col].fillna('None')
    
    # Select only numerical features for modeling
    feature_columns = [col for col in numeric_features if col != 'SalePrice']
    
    # Split back into train and test
    train_processed = all_data[:len(train_df)][feature_columns]
    test_processed = all_data[len(train_df):][feature_columns]
    
    return train_processed, test_processed

# Preprocess data
X_train, X_test = preprocess_data(train_fe, test_fe)
y_train = np.log1p(train_fe['SalePrice'])  # Log transform target

print(f"Training features shape: {X_train.shape}")
print(f"Test features shape: {X_test.shape}")
```

Выбираем модель из 6 претендентов с механизмом кросс-валидации

```python
# Cross-validation setup
kfold = KFold(n_splits=5, shuffle=True, random_state=42)

def evaluate_model(model, X, y, cv=kfold):
    """Evaluate model using cross-validation"""
    scores = cross_val_score(model, X, y, cv=cv, scoring='neg_root_mean_squared_error')
    return -scores.mean(), scores.std()

# Models to compare
models = {
    'Linear': LinearRegression(),
    'Ridge': Ridge(alpha=10),
    'Lasso': Lasso(alpha=0.001),
    'ElasticNet': ElasticNet(alpha=0.001, l1_ratio=0.5),
    'Random Forest': RandomForestRegressor(n_estimators=50, random_state=42),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=50, random_state=42)
}

# Compare models
results = {}
print("Model Comparison (Cross-Validation RMSE):")
print("-" * 45)

for name, model in models.items():
    mean_rmse, std_rmse = evaluate_model(model, X_train, y_train)
    results[name] = mean_rmse
    print(f"{name:<15}: {mean_rmse:.4f} (+/- {std_rmse*2:.4f})")

# Find best model
best_model_name = min(results, key=results.get)
print(f"\nBest model: {best_model_name} with RMSE: {results[best_model_name]:.4f}")
```
